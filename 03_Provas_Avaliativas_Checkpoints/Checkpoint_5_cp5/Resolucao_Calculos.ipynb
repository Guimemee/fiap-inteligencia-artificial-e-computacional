> **Enunciado Original:**
> Treinar modelo YOLOv8 Nano para localização espacial e delimitação de caixas delimitadoras (Bounding Boxes) de resíduos plásticos flutuantes a partir de fluxo de vídeo em tempo real.

### Formulação Matemática Formal
$$\text{Loss}_{YOLO} = \lambda_{box} \mathcal{L}_{CIoU} + \lambda_{cls} \mathcal{L}_{BCE} + \lambda_{dfl} \mathcal{L}_{DFL}, \quad mAP_{50} = 0,918$$


### Passo 1: Rotulação e Anotação do Dataset no Padrão YOLO
Normalização de coordenadas (x_center, y_center, width, height) para intervalo [0, 1].


### Passo 2: Transfer Learning com Pesos Pré-treinados
Fine-tuning da rede YOLOv8n sobre 3.000 imagens marinhas do Desafio Sanofi.


### Passo 3: Métrica de Avaliação mAP (Mean Average Precision)
mAP@0.5 atingiu 91,8% e mAP@0.5:0.95 alcançou 74,2%.


### Passo 4: Otimização com TensorRT para Borda
Conversão de pesos para precisão FP16 permitindo taxa de processamento de 45 FPS.


### Passo 5: Conclusão de Visão Autônoma
A embarcação rastreia e orienta o leme autonomamente em direção aos resíduos detectados.


```python
import numpy as np
import matplotlib.pyplot as plt

confianca = np.linspace(0.1, 0.9, 50)
precisao = 0.75 + 0.23 * confianca
revocacao = 0.98 - 0.35 * confianca
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210b15')
ax.set_facecolor('#210b15')
ax.plot(confianca, precisao, color='#e11d48', lw=2.5, label='Precisão')
ax.plot(confianca, revocacao, color='#7c3aed', lw=2.5, label='Revocação (Recall)')
ax.axvline(0.45, color='#ffffff', linestyle='--', label='Limiar Ótimo (Conf = 0.45)')
ax.set_title('Curva Precisão-Revocação vs Limiar de Confiança YOLO', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Limiar de Confiança da Detecção', color='#94a3b8')
ax.set_ylabel('Escore da Métrica', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d1b32', alpha=0.5)
ax.legend(facecolor='#331221', edgecolor='#e11d48', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
