### Formulação Matemática Formal
$$Z = \frac{X - \mu}{\sigma}, \quad X_{scaled} \sim \mathcal{N}(0, 1)$$


### Passo 1: Análise Exploratória e Matriz de Correlação
Identificação de redundâncias e eliminação de variáveis com variância próxima de zero.


### Passo 2: Imputação Inteligente com K-Nearest Neighbors (KNN)
Reconstrução de leituras faltantes com k=5 vizinhos mais próximos no espaço vetorial.


### Passo 3: Codificação de Atributos Categóricos
OneHotEncoder com drop='first' para prevenção de multicolinearidade matemática.


### Passo 4: Padronização Escalar (StandardScaler)
Transformação dos dados garantindo média zero e desvio padrão unitário para convergência de gradientes.


### Passo 5: Conclusão de Pipeline
Pipeline serializado com Joblib integrado ao ecossistema Scikit-Learn.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
bruto = np.random.exponential(5, 500)
padronizado = (bruto - np.mean(bruto)) / np.std(bruto)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210b15')
ax.set_facecolor('#210b15')
ax.hist(padronizado, bins=30, color='#e11d48', alpha=0.8, edgecolor='#331221')
ax.set_title('Distribuição Padronizada das Features (Z-Score)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Valor Normalizado (Média = 0, Desvio = 1)', color='#94a3b8')
ax.set_ylabel('Frequência de Amostras', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#4d1b32', alpha=0.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
