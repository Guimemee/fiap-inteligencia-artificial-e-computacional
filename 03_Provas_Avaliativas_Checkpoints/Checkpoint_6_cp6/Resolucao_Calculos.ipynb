```python
import numpy as np
import matplotlib.pyplot as plt

metodos = ['Navegação Aleatória', 'Monitoramento Estático', 'Roteamento com IA Preditiva']
residuos_kg = [42, 68, 145]
cores = ['#94a3b8', '#7c3aed', '#e11d48']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210b15')
ax.set_facecolor('#210b15')
bars = ax.bar(metodos, residuos_kg, color=cores, width=0.45)
ax.set_title('Coleta de Resíduos Plásticos por Missão (kg)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Massa Recolhida (kg)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#4d1b32', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 3, f'{yval} kg', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
