**Título:** Redes Neurais Convolucionais (CNN) para Classificação de Imagens

> **Enunciado Original:**
> Implementar arquitetura CNN com camadas Conv2D, MaxPooling2D e BatchNormalization para classificar tipos de fragmentos plásticos marinhos coletados por câmeras subaquáticas.

### Formulação Matemática Formal
$$(I * K)(i, j) = \sum_m \sum_n I(i - m, j - n) K(m, n), \quad \text{Pooling: } \max_{m, n} I(2i + m, 2j + n)$$


### Passo 1: Aumento de Dados (Data Augmentation)
Rotações aleatórias, espelhamento horizontal e zoom prevenindo memorização de fundo d'água.


### Passo 2: Blocos Convolucionais Hierárquicos
3 blocos de convolução (filtros 32, 64, 128 com kernel 3x3) seguidos de MaxPooling 2x2.


### Passo 3: Camada de Achatamento (Flatten) e Densa Final
Classificador Denso com ativação Softmax para 4 classes: PET, PEAD, Polipropileno e Matéria Orgânica.


### Passo 4: Matriz de Confusão nos Dados de Validação
Acurácia global de 96,4% com sensibilidade de 98,1% na detecção de plásticos nocivos.


### Passo 5: Conclusão de Visão Subaquática
Inferência realizada em 18 ms por quadro em aceleradores de hardware embarcados.


```python
import numpy as np
import matplotlib.pyplot as plt

classes = ['PET', 'PEAD', 'Polipropileno', 'Orgânico']
acuracia = [97.2, 95.8, 96.4, 98.1]
cores = ['#e11d48', '#7c3aed', '#f43f5e', '#10b981']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210b15')
ax.set_facecolor('#210b15')
bars = ax.bar(classes, acuracia, color=cores, width=0.5)
ax.set_ylim(80, 102)
ax.set_title('Acurácia por Classe de Resíduo Plástico (CNN)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Acurácia (%)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#4d1b32', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 0.8, f'{yval}%', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
