> **Enunciado Original:**
> Reduzir 16 dimensões de parâmetros oceânicos para os 2 primeiros Componentes Principais (PCA) e aplicar o algoritmo K-Means determinando o número ótimo de clusters pelo Método do Cotovelo.

### Formulação Matemática Formal
$$\text{Autovalores}: \det(\Sigma - \lambda I) = 0, \quad \text{Variância Explicada} = \frac{\lambda_1 + \lambda_2}{\sum \lambda_i} > 85\%$$


### Passo 1: Matriz de Covariância dos Dados Padronizados
\Sigma = \frac{1}{n-1} X^T X de dimensão 16 \times 16.


### Passo 2: Decomposição Espectral em Autovetores e Autovalores
Ordenação decrescente dos autovalores para identificar os eixos de maior variância.


### Passo 3: Projeção no Espaço Bidimensional
PC_1 e PC_2 retêm sozinhos 86,4% da informação total do ecossistema marinho.


### Passo 4: Método do Cotovelo (Elbow Method) para K-Means
Cálculo da inércia (WCSS) para K de 1 a 8, revelando inflexão ótima em K = 3.


### Passo 5: Conclusão de Agrupamento
Identificação clara de 3 zonas oceânicas: Prístina, Poluição Moderada e Ponto Crítico de Microplásticos.


```python
import numpy as np
import matplotlib.pyplot as plt

k = np.arange(1, 9)
wcss = [1850, 920, 410, 320, 260, 220, 190, 170]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210b15')
ax.set_facecolor('#210b15')
ax.plot(k, wcss, 'o-', color='#e11d48', lw=2.5, label='Inércia Residual (WCSS)')
ax.scatter([3], [410], color='#7c3aed', s=90, zorder=5, label='Ponto do Cotovelo (K = 3)')
ax.set_title('Método do Cotovelo: Determinação Ótima de K', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Número de Clusters (K)', color='#94a3b8')
ax.set_ylabel('Soma dos Quadrados Intra-Cluster (WCSS)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d1b32', alpha=0.5)
ax.legend(facecolor='#331221', edgecolor='#e11d48', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
