**Título:** Redes Neurais Artificiais (MLP) com Keras e Otimizador Adam

> **Enunciado Original:**
> Construir e treinar um Perceptron Multicamadas (MLP) no Keras com camadas densas, regularização Dropout e função de ativação ReLU para prever a concentração de microplásticos.

### Formulação Matemática Formal
$$z^{[l]} = W^{[l]} a^{[l-1]} + b^{[l]}, \quad a^{[l]} = \text{ReLU}(z^{[l]}) = \max(0, z^{[l]})$$


### Passo 1: Arquitetura da Rede
Camada de Entrada (16 neurônios), Camada Oculta 1 (64 neurônios, ReLU), Dropout(0,3), Camada Oculta 2 (32 neurônios, ReLU), Saída (1 neurônio).


### Passo 2: Função de Custo e Otimizador
Mean Squared Error (MSE) com otimizador Adam (taxa de aprendizado inicial \alpha = 0,001).


### Passo 3: Callbacks de Treinamento
EarlyStopping monitorando val_loss com paciência de 15 épocas e ModelCheckpoint salvando os melhores pesos.


### Passo 4: Curvas de Aprendizado
Convergência suave sem indícios de sobreajuste (overfitting).


### Passo 5: Conclusão Numérica
R^2 = 0,948 no conjunto de teste e erro absoluto médio MAE inferior a 0,12 mg/L.


```python
import numpy as np
import matplotlib.pyplot as plt

epocas = np.arange(1, 41)
train_loss = 0.8 * np.exp(-epocas / 8) + 0.03
val_loss = 0.85 * np.exp(-epocas / 8) + 0.045 + np.random.normal(0, 0.005, 40)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210b15')
ax.set_facecolor('#210b15')
ax.plot(epocas, train_loss, color='#e11d48', lw=2.5, label='Loss de Treino (MSE)')
ax.plot(epocas, val_loss, color='#7c3aed', lw=2, linestyle='--', label='Loss de Validação (val_loss)')
ax.set_title('Convergência da Rede Neural MLP no Keras', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Épocas de Treinamento', color='#94a3b8')
ax.set_ylabel('Erro Médio Quadrático (MSE)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d1b32', alpha=0.5)
ax.legend(facecolor='#331221', edgecolor='#e11d48', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
