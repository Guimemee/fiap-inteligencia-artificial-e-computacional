## EXERCÍCIO 04

Crie uma estrutura de repetição para fazer a leitura de cinco números inteiros e os armazene dentro de uma lista. Após a leitura, crie outra estrutura de repetição para somar os valores digitados.





````python
# prompt: Crie uma estrutura de repetição para fazer a leitura de cinco números inteiros e os armazene dentro de uma lista. Após a leitura, crie outra estrutura de repetição para somar os valores digitados.

# Criando uma lista vazia para armazenar os números inteiros
lista = []

# Criando um loop para ler cinco números inteiros
for i in range(5):
  # Lendo um número inteiro
  numero = int(input("Digite um número inteiro: "))

  # Adicionando o número inteiro à lista
  lista.append(numero)

# Criando uma variável para armazenar a soma dos números inteiros
soma = 0

# Criando um loop para somar os números inteiros da lista
for numero in lista:
  # Somando o número inteiro à variável soma
  soma += numero

# Imprimindo a soma dos números inteiros
print("A soma dos números inteiros é:", soma)
````
